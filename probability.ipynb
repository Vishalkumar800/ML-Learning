````python
import pandas as pd
import random
import numpy as np

L = []
for i in range(10000):
  L.append(random.randint(1,6))

data = pd.Series(L)
````

````python
# find probablility

prob = (data.value_counts() / 10000).sort_index() # yha data pta hai kitne baar hai agr real data ho toh

# data.value_counts().sum() yeah apka total count krke bateyga
````

````python
prob.plot(kind="bar")
````

# Let suppose 2 dice ek sath and sum ka prob pucha jayee

````python
L = []

for i in range(10000):
  a = random.randint(1,6)
  b = random.randint(1,6)

  c = a +b
  L.append(c)

data = pd.Series(L)
````

````python
twodiceProb = data.value_counts() / data.value_counts().sum()
twodiceProb.sort_index(inplace=True)
````

````python
twodiceProb.plot(kind="bar")
````

````python
# CDF (cummulative gives you probaility up to x)

A = []

for i in range(10000):
  A.append(random.randint(1,6))

A[:5]

data = pd.Series(A)
````

````python
pb = (data.value_counts() / data.value_counts().sum()).sort_index()
````

````python
# PDF  esme batata hai ek particular value ki probability like if i said 4 toh 4 ki probality in data

pb.plot(kind="bar")
````

````python
# CDF

d = np.cumsum(pb)
````

````python
d.plot(kind="bar") #yeah graph CMF ka hai using PDF
# jisme agr me 4 bolu toh uski probility of 4 aur ussse less means 1 to 4 combine aa rhi hai.
````

# Normal Distribution function (PDF Probablity density function )

````python
import numpy as np
from numpy.random import normal
import matplotlib.pyplot as plt

sample = normal(loc = 50 , scale = 5 , size = 10000)
````

````python
sample
````

````python
plt.hist(sample, bins = 10
)
````

````python
# graph kuch thoda sa normal distribution ki tarah hai
# calculate sample mean and sample deviation

sample_mean = sample.mean()
sample_std = sample.std()
````

````python
from scipy.stats import norm
dist = norm(sample_mean , sample_std)
````

NORMAL DISTRIBUTION — scipy.stats.norm

Import:
from scipy.stats import norm

Create Normal Distribution:
dist = norm(sample_mean, sample_std)

Meaning:
norm(mean, standard_deviation)

sample_mean → Distribution ka center/mean
sample_std  → Distribution ka spread/standard deviation

Example:
sample_mean = 50
sample_std = 5

dist = norm(sample_mean, sample_std)

# mean = 50

# standard deviation = 5

Important:
dist ek Normal Distribution object hai.

PDF:
dist.pdf(x)
→ x par probability density deta hai.

CDF:
dist.cdf(x)
→ P(X <= x), yani x ke left side ki cumulative probability.

PPF:
dist.ppf(p)
→ Given probability p ke liye x value deta hai.


````python
values = np.linspace(sample.min() , sample.max(),100)
````

````python
prob = [dist.pdf(value) for value in values ]
````

NumPy ka `hstack()` ka matlab hai **arrays ko horizontally (side-by-side) jodna**.

```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

result = np.hstack((a, b))

print(result)
```

Output:

```text
[1 2 3 4 5 6]
```

### 2D example

```python
a = np.array([[1, 2],
              [3, 4]])

b = np.array([[5, 6],
              [7, 8]])

np.hstack((a, b))
```

Output:

```text
[[1 2 5 6]
 [3 4 7 8]]
```

Yaani:

```text
a          b
[1 2]  +  [5 6]  → [1 2 5 6]
[3 4]     [7 8]     [3 4 7 8]
```

### Yaad rakhne ka shortcut 🧠

* `hstack()` → **Horizontal** → side by side → columns increase
* `vstack()` → **Vertical** → upar-niche → rows increase

```python
np.hstack((a, b))
np.vstack((a, b))
```
