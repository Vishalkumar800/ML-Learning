```python
import pandas as pd
import random

L = []
for i in range(10000):
  L.append(random.randint(1,6))

data = pd.Series(L)
```

```python
# find probablility

prob = (data.value_counts() / 10000).sort_index() # yha data pta hai kitne baar hai agr real data ho toh

# data.value_counts().sum() yeah apka total count krke bateyga
```

```python
prob.plot(kind="bar")
```

# Let suppose 2 dice ek sath and sum ka prob pucha jayee

```python
L = []

for i in range(10000):
  a = random.randint(1,6)
  b = random.randint(1,6)

  c = a +b
  L.append(c)

data = pd.Series(L)
```

```python
twodiceProb = data.value_counts() / data.value_counts().sum()
twodiceProb.sort_index(inplace=True)
```

```python
twodiceProb.plot(kind="bar")
```
